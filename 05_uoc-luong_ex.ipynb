```python
# a
means20 = []

for i in range(100):
    sample = np.random.choice(POP, size=20, replace=False)
    mean_sample = np.mean(sample)
    means20.append(mean_sample)
    
plt.figure(figsize=(8, 5))

plt.hist(means20, bins=20, edgecolor='black')

plt.xlabel("Trung bình mẫu")
plt.ylabel("Tần số")
plt.title("Phân phối của 100 trung bình mẫu - n = 20")

plt.show()
# b
means100 = []

for i in range(100):
    sample = np.random.choice(POP, size=100, replace=False)
    mean_sample = np.mean(sample)
    means100.append(mean_sample)
    
plt.figure(figsize=(8, 5))

plt.hist(means100, bins=20, edgecolor='black')

plt.xlabel("Trung bình mẫu")
plt.ylabel("Tần số")
plt.title("Phân phối của 100 trung bình mẫu - n = 100")

plt.show()
# c
plt.figure(figsize=(10, 5))

plt.hist(
    means20,
    bins=20,
    alpha=0.6,
    label="n = 20"
)

plt.hist(
    means100,
    bins=20,
    alpha=0.6,
    label="n = 100"
)

plt.xlabel("Trung bình mẫu")
plt.ylabel("Tần số")
plt.title("So sánh phân phối của trung bình mẫu")

plt.legend()
plt.show()
```

### Bài 6: Khảo sát phương sai mẫu

Bạn hãy mô phỏng 1000 lần lấy mẫu có kích cỡ 100. Vẽ histogram đại lượng phương sai của các mẫu (Phương sai bằng bình phương độ lệch chuẩn)

```python
phuong_sai_100 = []

for i in range(1000):
    sample100 = np.random.choice(POP, size=100, replace=False)
    p = np.var(sample100,ddof=1 )
    phuong_sai_100.append(p)
plt.figure(figsize=(8, 5))

plt.hist(phuong_sai_100, bins=20, edgecolor='black')

plt.xlabel("Phương sai mẫu")
plt.ylabel("Tần số")
plt.title("Phân phối của phương sai mẫu")

plt.show()

print("Phương sai trung bình:", np.mean(phuong_sai_100))
print("Phương sai quần thể:", np.var(POP))
```

### Bài 7:  Bài Tập Ước Lượng 

Khảo sát cân nặng (kg) của gà khi xuất chuồng, người ta cân một số con và có kết quả như sau:

> 2.1, 1.8, 2.0, 2.3, 1.7, 1.5, 2.0, 2.2, 1.8

Giả sử cân nặng của gà là biến ngẫu nhiên có phân phối chuẩn. Hãy lập khoảng tin cậy cho cân nặng trung bình của  khi xuất chuồng với độ tin cậy là 95%. Trong 2 trường hợp:

a, Biết $\sigma = 0.3$

b, Không biết $\sigma$


```python
#cau7
data = np.array([2.1, 1.8, 2.0, 2.3, 1.7, 1.5, 2.0, 2.2, 1.8])

n = len(data)
mean = np.mean(data)

print("Trung bình mẫu:", mean)
#a. Biết σ = 0.3
#Biết độ lệch chuẩn quần thể → dùng Z:

sigma = 0.3
confidence = 0.95

z = stats.norm.ppf(1 - (1-confidence)/2)

error = z * sigma / np.sqrt(n)

lower = mean - error
upper = mean + error

print("Khoảng tin cậy 95%:", (lower, upper))

#b. Không biết σ
#Không biết độ lệch chuẩn quần thể → dùng t-Student:

s = np.std(data, ddof=1)

t = stats.t.ppf(1 - (1-confidence)/2, n-1)

error = t * s / np.sqrt(n)

lower = mean - error
upper = mean + error

print("Độ lệch chuẩn mẫu:", s)
print("Khoảng tin cậy 95%:", (lower, upper))
```
